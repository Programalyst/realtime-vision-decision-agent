```python
import torch

print("PyTorch Version:", torch.__version__)

if torch.backends.mps.is_available():
    print("✅ Success! Found Apple Silicon GPU (MPS Mode Enabled)")
    device_target = "mps"
else:
    print("⚠️ MPS acceleration not found. Defaulting to CPU.")
    device_target = "cpu"
    
```

```python
from roboflow import Roboflow
from ultralytics import YOLO

# 1. Replace this block with the code snippet from Roboflow
# Go to "Versions" > "Download Dataset" > select YOLOv11 > "Show download code"
rf = Roboflow(api_key="")
project = rf.workspace("leonard-lin-yrqtn").project("collect-raindrops")
version = project.version(3)
dataset = version.download("yolov11")
                
# 2. Initialize the lightweight YOLOv11 nano framework
model = YOLO("yolo11n.pt") 

# 3. Kick off local training using Apple Silicon acceleration
results = model.train(
    data=f"{dataset.location}/data.yaml", 
    epochs=150,
    patience=30,
    batch=16,
    imgsz=640,             
    device=device_target,
    val=True,
)
```
